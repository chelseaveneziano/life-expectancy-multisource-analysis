Step #1: Scraping the data

```python
import pandas as pd

# read the HTML tables from the website

url = "https://en.wikipedia.org/wiki/List_of_countries_by_life_expectancy"
tables = pd.read_html(url)

# display the number of tables in the dataset and inspect their content
print(f"Number of tables found: {len(tables)}")

# display the first few rows of each table sto find the correct one
for i, table in enumerate(tables):
    print(f"Table {i}:")
    print(table.head())
    print("\n")
```

Explantation: This code reads all HTML tables from the website and prints the first few rows of each table, so I can visually inspect and choose the correct ones.

Step 4: Flatten Multiindex Columns and Display Columns

```python
# function to flatten MultiIndex columns for tables 1, 2 and 3
def flatten_columns(df, table_name):
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [' '.join(col).strip() for col in df.columns.values]
        print(f"MultiIndex columns flattened for {table_name}")
    else:
        print(f"No MultiIndex found in {table_name}")
        
# flatten the columns for tables with MultiIndex
flatten_columns(df1, 'Table 1')
flatten_columns(df2, 'Table 2')
flatten_columns(df3, 'Table 3')

# print the flattened columns
print("\nTable 1 columns:")
print(df1.columns)
print("\nTable 2 columns:")
print(df2.columns)
print("\nTable 3 columns:")
print(df3.columns)
```

Explanation: This code flattens the MultiIndex columns in df1, df2, and df3 by joining the column levels with a space. It then prints the columns for verification to ensure they are properly flattened.

Step # 6: Check for extra spaces in column names

```python
# define a function to check for extra spaces in the column names
def check_column_spaces(df, table_name):
    for col in df.columns:
        if ' ' in ''.join(col):
            print(f"Extra spaces found in column: '{col}' in {table_name}")

# apply the check to each table
tables_to_check = {
    'Table 1': df1,
    'Table 2': df2,
    'Table 3': df3,
    'Table 5': df5
}

# iterate through each table and check for extra spaces
for table_name, df in tables_to_check.items():
    print(f"\nChecking {table_name} for extra spaces in column names:")
    check_column_spaces(df, table_name)
    
```

Explanation: This code will help me identify whether extra spaces are present in any of the tables column names.

Step # 7: Remove extra spaces from column names

```python
# define a function to remove extra spaces from column names
def remove_column_spaces(df, table_name):
    df.columns = [col.strip() for col in df.columns]
    print(f"Extra spaces removed from columns names in {table_name}")

# apply the function to each table that has extra spaces
tables_to_fix = {
    'Table 1': df1,
    'Table 2': df2,
    'Table 3': df3,
    'Table 5': df5
}

# iterate through each table and remove extra spaces
for table_name, df in tables_to_fix.items():
    print(f"\nFixing {table_name}:")
    remove_column_spaces(df, table_name)

# display the updated column names
for table_name, df in tables_to_fix.items():
    print(df.columns)
```

Explanation: This code removes extra spaces from the column names in each table and then shows the updated names to check if they are fixed.

Step 10: Convert Relevant Columns to Numeric

```python
def convert_to_numeric(df, numeric_columns):
    for col in numeric_columns:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce') # convert and handle errors
            
# apply conversion to relevant numeric columns 
# for df1 (21 columns)
numeric_columns_df1 = ['Life Expectancy Both Sexes at Birth', 'Life Expectancy Bonus 0→15', 'Life Expectancy at 15',
                       'Life Expectancy Bonus 15→65', 'Life Expectancy at 65', 'Life Expectancy Bonus 65→80',
                       'Life Expectancy at 80', 'Male Life Expectancy at Birth', 'Male Life Expectancy at 15',
                       'Male Life Expectancy at 65', 'Male Life Expectancy at 80', 'Female Life Expectancy at Birth',
                       'Female Life Expectancy at 15', 'Female Life Expectancy at 65', 'Female Life Expectancy at 80',
                       'Sex Gap at Birth', 'Sex Gap at 15', 'Sex Gap at 65', 'Sex Gap at 80']
convert_to_numeric(df1, numeric_columns_df1)

# for df2 (16 columns)
numeric_columns_df2 = ['2023 Life Expectancy All', '2023 Male', '2023 Female', '2023 Sex Gap', 
                       'Historical Life Expectancy 2019', '2019 → 2020 Change', 'Historical Life Expectancy 2020',
                       '2020 → 2021 Change', 'Historical Life Expectancy 2021', '2021 → 2022 Change', 
                       'Historical Life Expectancy 2022', '2022 → 2023 Change', 'Historical Life Expectancy 2023']
convert_to_numeric(df2, numeric_columns_df2)

# For df3 (16 columns)
numeric_columns_df3 = ['2022 Life Expectancy All', '2022 Male', '2022 Female', '2022 Sex Gap',
                       'Historical Life Expectancy 2014', '2014 → 2019 Change', 'Historical Life Expectancy 2019',
                       '2019 → 2020 Change', 'Historical Life Expectancy 2020', '2020 → 2021 Change',
                       'Historical Life Expectancy 2021', '2021 → 2022 Change', 'Historical Life Expectancy 2022']
convert_to_numeric(df3, numeric_columns_df3)

# For df5 (5 columns)
numeric_columns_df5 = ['Life Expectancy All', 'Male', 'Female', 'Sex Gap']
convert_to_numeric(df5, numeric_columns_df5)

# print conversion
print(df1.dtypes)
print(df2.dtypes)
print(df3.dtypes)
print(df5.dtypes)
```

Explanation: This step converts the relevant columns to numeric format using pd.to_numeric(), ensuring that any invalid values are handled by converting non-numeric entries into NaN for easier processing.

Step 11: Handle missing values

```python
# fill the missing values with the median for numeric columns
def handle_missing_values(df, numeric_columns):
    for col in numeric_columns:
        if col in df.columns:
            df[col].fillna(df[col].median(), inplace=True) # fill NAN with median
        else:
            print(f"Column '{col}' not found in DataFrame")

# apply handling of missing values to numeric columns

handle_missing_values(df1, numeric_columns_df1)
handle_missing_values(df2, numeric_columns_df2)
handle_missing_values(df3, numeric_columns_df3)
handle_missing_values(df5, numeric_columns_df5)
```

Explanation: This step fills missing values in the numeric columns with the median value, ensuring data consistency for further analysis.

Step 12: Removing unnecessary columns if they exist

```python
def remove_unnecessary_columns(df, columns_to_remove):
    for col in columns_to_remove:
        if col in df.columns:
            df.drop(columns=col, inplace=True)
        else:
            print(f"Column '{col}' not found in DataFrame.")

# apply to each dataframe
remove_unnecessary_columns(df1, ['Unnamed'])
remove_unnecessary_columns(df2, ['Unnamed'])
remove_unnecessary_columns(df3, ['Unnamed'])
```

Explanation: I used this code as a final check to confirm that the dataset is clean and that no irrelevant columns remain, ensuring consistency across all DataFrames before analysis.
